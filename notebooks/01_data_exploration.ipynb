# Notebook 1 : Exploration des Données
Chargement, statistiques descriptives et analyse exploratoire du dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/student_data.csv')
print('Shape:', df.shape)
print('\n--- Premières lignes ---')
df.head()
```

```python
print('--- Statistiques descriptives ---')
df.describe().T[['mean', 'std', 'min', '50%', 'max']]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Distribution de la cible
sns.countplot(x='academic_success', data=df, palette='viridis', ax=axes[0])
axes[0].set_title('Distribution de la Cible (0=Non-réussite, 1=Réussite)')
axes[0].set_xlabel('Réussite Académique')
axes[0].set_ylabel('Effectif')

# Matrice de corrélation
sns.heatmap(df.corr(numeric_only=True), annot=False, cmap='coolwarm', ax=axes[1], center=0)
axes[1].set_title('Matrice de Corrélation')

plt.tight_layout()
plt.show()
```

```python
print('--- Taux de réussite ---')
print(f"{df['academic_success'].mean()*100:.1f}% des étudiants réussissent")

print('\n--- Valeurs manquantes ---')
print(df.isnull().sum().sum())
```

### Conclusion
Le dataset contient 395 étudiants avec 15 variables et 1 cible binaire. Les données sont synthétiques, générées avec les mêmes distributions que le dataset UCI Student Performance.

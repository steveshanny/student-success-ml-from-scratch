# Notebook 2 : Analyse en Composantes Principales (ACP)
Implémentation from scratch de l'ACP avec décomposition spectrale.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src.preprocessing import StandardScalerScratch, train_test_split_scratch

df = pd.read_csv('../data/raw/student_data.csv')
X = df.drop(columns=['academic_success']).values
y = df['academic_success'].values

X_train, X_test, y_train, y_test = train_test_split_scratch(X, y, test_size=0.2, random_state=42)
scaler = StandardScalerScratch()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print('Données standardisées. Shape train :', X_train_scaled.shape)
```

## Implémentation ACP From Scratch

**Étape 1 : Matrice de covariance empirique :**
$$\Sigma = \frac{1}{m} X^T X$$

**Étape 2 : Décomposition spectrale (eigenvalues/vectors) :**
$$\Sigma v_i = \lambda_i v_i$$

**Étape 3 : Projection :**
$$Z = X W$$

```python
from src.pca_scratch import PCAFromScratch

pca = PCAFromScratch(n_components=2)
Z_train = pca.fit_transform(X_train_scaled)

print('Valeurs propres :', pca.eigenvalues_)
print('Ratio variance expliquée (PC1, PC2) :', pca.explained_variance_ratio_)
print('Variance cumulée 2 composantes :', sum(pca.explained_variance_ratio_))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scatter 2D
scatter = axes[0].scatter(Z_train[:, 0], Z_train[:, 1], c=y_train, cmap='coolwarm', alpha=0.7, edgecolors='k')
axes[0].set_xlabel('PC1')
axes[0].set_ylabel('PC2')
axes[0].set_title('Projection ACP 2D des Étudiants')
fig.colorbar(scatter, ax=axes[0], label='Réussite')

# Scree plot
full_pca = PCAFromScratch(n_components=X_train_scaled.shape[1])
full_pca.fit(X_train_scaled)
cum_var = np.cumsum(full_pca.explained_variance_ratio_)
axes[1].bar(range(1, len(cum_var)+1), full_pca.explained_variance_ratio_, alpha=0.6, label='Individuelle')
axes[1].step(range(1, len(cum_var)+1), cum_var, where='mid', color='red', label='Cumulée')
axes[1].set_xlabel('Composantes')
axes[1].set_ylabel('Variance Expliquée')
axes[1].set_title('Scree Plot')
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Conclusion
L'ACP from scratch réduit la dimensionnalité tout en préservant la majorité de la variance. Les deux premières composantes captrent l'essentiel de l'information.

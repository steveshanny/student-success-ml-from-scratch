# Notebook 3 : Régression Logistique From Scratch
Descente de gradient vectorisée avec Log-Loss et régularisation L2.

```python
import numpy as np
import matplotlib.pyplot as plt

from src.preprocessing import StandardScalerScratch, train_test_split_scratch

import pandas as pd
df = pd.read_csv('../data/raw/student_data.csv')
X = df.drop(columns=['academic_success']).values
y = df['academic_success'].values

X_train, X_test, y_train, y_test = train_test_split_scratch(X, y, test_size=0.2, random_state=42)
scaler = StandardScalerScratch()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Fonction de Coût Log-Loss + L2

$$J(\theta) = -\frac{1}{m} \sum [ y \log(\hat{y}) + (1-y)\log(1-\hat{y}) ] + \frac{\lambda}{2m} \sum_{j=1}^n \theta_j^2$$

**Gradient :**
$$\nabla J(\theta) = \frac{1}{m} X^T (\hat{y} - y) + \frac{\lambda}{m} \theta$$

```python
from src.logistic_regression_scratch import LogisticRegressionScratch

model = LogisticRegressionScratch(learning_rate=0.1, l2_lambda=0.1, n_iterations=1000)
model.fit(X_train_scaled, y_train)

plt.figure(figsize=(8, 5))
plt.plot(model.cost_history, color='blue', linewidth=2)
plt.title(f'Convergence (Coût initial: {model.cost_history[0]:.4f} -> Final: {model.cost_history[-1]:.4f})')
plt.xlabel('Itérations')
plt.ylabel('Coût Log-Loss + L2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## Impact du Learning Rate

```python
alphas = [0.001, 0.1, 3.5]
plt.figure(figsize=(8, 5))

for alpha in alphas:
    m = LogisticRegressionScratch(learning_rate=alpha, l2_lambda=0.01, n_iterations=200)
    m.fit(X_train_scaled, y_train)
    plt.plot(m.cost_history, label=f'alpha = {alpha}')

plt.xlabel('Itérations')
plt.ylabel('Coût J(theta)')
plt.title('Impact du Learning Rate sur la Convergence')
plt.ylim(0, 1.5)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
from src.metrics import accuracy_score_scratch, f1_score_scratch, confusion_matrix_scratch
import seaborn as sns

preds = model.predict(X_test_scaled)
probs = model.predict_proba(X_test_scaled)

print(f'Accuracy Test : {accuracy_score_scratch(y_test, preds)*100:.2f}%')
print(f'F1-Score Test : {f1_score_scratch(y_test, preds)*100:.2f}%')

cm = confusion_matrix_scratch(y_test, preds)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Non-réussite', 'Réussite'],
            yticklabels=['Non-réussite', 'Réussite'])
plt.xlabel('Prédit')
plt.ylabel('Réel')
plt.title('Matrice de Confusion')
plt.show()
```

### Conclusion
La régression logistique from scratch avec régularisation L2 converge correctement. Le choix du learning rate est crucial pour la stabilité de la descente de gradient.
